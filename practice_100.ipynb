```python
class Circle:
  def __init__(self, radius):
    self.radius = radius

  def area(self):
    return "Area:", (22/7 * self.radius)

  def circumference(self):
    return "Circumference:", (2*(22/7)*self.radius)

c = Circle(2)
print(c.area())
print(c.circumference())
```

```python
class Rectangle:
  def __init__(self, length, breadth):
    self.l = length
    self.b = breadth

  def area(self):
    return (self.l*self.b)

  def peri(self):
    return (2*(self.l+self.b))

r = Rectangle(4,5)
print("Area: ", r.area())
print("Perimeter: ", r.peri())
```

```python
class Marks:
  def __init__(self, sub1, sub2, sub3, sub4, sub5):
    self.s1 = sub1
    self.s2 = sub2
    self.s3 = sub3
    self.s4 = sub4
    self.s5 = sub5

  def total(self):
    return self.s1+self.s2+self.s3+self.s4+self.s5

  def avg(self):
    return (self.total()/5)

m = Marks(95,96,97,98,99)
print(m.total())
print(m.avg())
```

```python
class Time:
  def __init__(self, seconds) -> None:
    self.s = seconds

  def hrs(self):
    return round(self.s/3600)

  def mins(self):
    return round(self.s/60)%60

  def secs(self):
    return round(self.s%60)

s1 = Time(15617)
print(s1.hrs(), s1.mins(), s1.secs())
```

```python
class EO:
  def __init__(self, num):
    self.num = num

  def what(self):
    if self.num%2 == 0:
      return "Even!"
    else:
      return "Odd!"

n = EO(9)
print(n.what())
```

```python
class What:
  def __init__(self, num) -> None:
    self.num = num

  def what(self):
    if self.num == 0:
      return "Zero"
    if self.num > 0:
      return "Positive"
    else:
      return "Negative"

n = What(-22)
print(n.what())
```

```python
class Largest:
  def __init__(self, n1, n2, n3) -> None:
    self.n1 = n1
    self.n2 = n2
    self.n3 = n3

  def lar(self):
    if self.n1 > self.n2 and self.n1 > self.n3:
      return self.n1
    if self.n2 > self.n1 and self.n2 > self.n3:
      return self.n2
    if self.n3 > self.n1 and self.n3 > self.n2:
      return self.n3

n = Largest(34, 49, 10)
print(n.lar())
```

```python
class Smallest:
  def __init__(self, n1, n2, n3) -> None:
    self.n1 = n1
    self.n2 = n2
    self.n3 = n3

  def sml(self):
    if self.n1 < self.n2 and self.n1 < self.n3:
      return self.n1
    if self.n2 < self.n1 and self.n2 < self.n3:
      return self.n2
    if self.n3 < self.n1 and self.n3 < self.n2:
      return self.n3

n = Smallest(34, 49, 10)
print(n.sml())
```

```python
class Sum:
  def __init__(self, n):
    self.n = n

  def sum(self):
    sum = 0
    for i in range(0, self.n+1):
      sum = sum + i
    return sum

s1 = Sum(10)
s1.sum()
```

```python
class Factorial:
  def __init__(self, n):
    self.n = n

  def product(self):
    product = 1
    for i in range(1, self.n+1):
      product = product * i
    return product

s1 = Factorial(5)
s1.product()
```

```python
class Count:
  def __init__(self, n):
    self.n = n

  def count(self):
    cnt = 0
    for i in range(1, self.n+1):
      if i%3 == 0:
        cnt += 1
    print(cnt)

n1 = Count(10)
n1.count()
```

```python
class Count_:
  def __init__(self, n):
    self.n = n

  def count_(self):
    cnt = 0
    for i in range(1, self.n+1):
      if i%3 == 0 or i%5 == 0:
        cnt += 1
    print(cnt)

n1 = Count_(10)
n1.count_()
```

```python
# Selection Sort
n = [5, 2, 8, 1, 10, 6]
for i in range(len(n)):
  min_index = i
  for j in range(i,len(n)):
    if n[j] < n[min_index]:
      min_index = j

  a = n[i]
  n[i] = n[min_index]
  n[min_index] = a
  print(n)

n
```

```python
# Bubble Sort
n = [5, 2, 8, 1, 10, 6]
for i in range(len(n)):
  something = False
  for j in range(len(n)-1-i):
    if n[j] > n[j+1]:
      a = n[j]
      n[j] = n[j+1]
      n[j+1] = a
      something = True

  if not something:
    break

n
```
