```python
#selectionSort

arr = [7, 2, 9, 5, 6, 10, 4, 1]

for i in range(len(arr)):
  min = i
  for j in range(i, len(arr)):
    if arr[min] > arr[j]:
      min = j

  arr[i], arr[min] = arr[min], arr[i]

arr
```

```python
#bubbleSort

arr = [7, 2, 9, 5, 6, 10, 4, 1]

for i in range(len(arr)):
  for j in range(i+1, len(arr)):
    if arr[i] > arr[j]:
      arr[i], arr[j] = arr[j], arr[i]

arr
```

```python
#bubbleSort
#optimised[O(n)]

arr = [1, 2, 4, 5, 6, 7, 9, 10]

swapped = False
for i in range(len(arr)):
  for j in range(i+1, len(arr)):
    if arr[i] > arr[j]:
      arr[i], arr[j] = arr[j], arr[i]
      swapped = True

  if not swapped:
    break

arr
```

```python
#insertionSort

arr = [7, 2, 9, 5, 6, 10, 4, 1]

for i in range(1, len(arr)):
  j = i
  while j > 0 and arr[j-1] > arr[j]: #the way we're moving across each array in every loop
    arr[j], arr[j-1] = arr[j-1], arr[j]
    j -= 1

arr
```

```python
#insertionSort
#Derived

arr = [7, 2, 9, 5, 6, 10, 4, 1]

for i in range(0, len(arr)):
  key = arr[i]
  j = i-1
  while j >= 0 and arr[j] > key:
    arr[j+1] = arr[j]
    j -= 1

  arr[j+1] = key

arr
```

```python
#mergeSort
#Derived

arr = [7, 2, 9, 5, 6, 10, 4, 1]

def merge_sort(arr):
  # base case
  if len(arr) == 1:
    return arr

  # divide
  mid = len(arr)//2
  left = arr[:mid]
  right = arr[mid:]

  # recurssion
  left = merge_sort(left)
  right = merge_sort(right)

  # merge
  i = 0 #left pointer
  j = 0 #right pointer
  l = []
  while i < len(left) and j < len(right):
    if left[i] < right[j]:
      l.append(left[i])
      i += 1
    else:
      l.append(right[j])
      j += 1

  # Copy the remaining elements from the left
  while i < len(left):
      l.append(left[i])
      i += 1

  # Copy the remaining elements from the right
  while j < len(right):
      l.append(right[j])
      j += 1

  return l
```

```python
#mergeSort

arr = [7, 2, 9, 5, 6, 10, 4, 1]

def merge(arr, low, mid, high):
  left_ptr = low #arr[low:mid+1]
  right_ptr = mid+1 #arr[mid+1:high+1]
  l = []

  while left_ptr <= mid and right_ptr <= high:
    if arr[left_ptr] < arr[right_ptr]:
      l.append(arr[left_ptr])
      left_ptr += 1
    else:
      l.append(arr[right_ptr])
      right_ptr += 1

  while left_ptr <= mid:
    l.append(arr[left_ptr])
    left_ptr += 1

  while right_ptr <= high:
    l.append(arr[right_ptr])
    right_ptr += 1

  for i in range(len(l)):
    arr[low+i] = l[i]


def merge_sort(arr, low, high):
  #base case
  if low >= high:
    return

  #divide
  mid = (low+high)//2

  #recursive calls
  merge_sort(arr, low, mid)
  merge_sort(arr, mid+1, high)

  #merge
  merge(arr, low, mid, high)


  return arr
```

```python
#quickSort
#Pseudo
def quick_sort(arr, low, high):
  #base case
  if low < high:
    return

  partition_idx = partition(arr, low, high)

  quick_sort(arr, low, partition_idx)
  quick_sort(arr, partition_idx + 1, high)


def partition(arr, low, high):
  pivot = arr[low]
  i = low
  j = high

  while i < j:
    while arr[i] < arr[pivot] and i <= high:
      i += 1 # find the first element greater than pivot

    while arr[j] > arr[pivot] and j >= low:
      j -= 1 # find the first element smaller than pivot

    if i < j:
      arr[low], arr[j] = arr[j], arr[low]
```
